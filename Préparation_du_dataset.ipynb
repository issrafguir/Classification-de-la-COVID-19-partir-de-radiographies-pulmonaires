```python
import os, cv2 , random
from tqdm._tqdm_notebook import tqdm_notebook as tqdm
# Vérification du nombre d'images par classe
covid_path = '/content/COVID-19_Radiography_Dataset/COVID/images'
normal_path = '/content/COVID-19_Radiography_Dataset/Normal/images'

covid_images = len([f for f in os.listdir(covid_path) if os.path.isfile(os.path.join(covid_path, f))])
normal_images = len([f for f in os.listdir(normal_path) if os.path.isfile(os.path.join(normal_path, f))])
print(f"Nombre d'images dans COVID : {covid_images}")
print(f"Nombre d'images dans Normal : {normal_images}")

```

```python
# Vérifier les noms de fichiers pour robustesse
covid_files = [f for f in os.listdir(covid_path) if os.path.isfile(os.path.join(covid_path, f))]
normal_files = [f for f in os.listdir(normal_path) if os.path.isfile(os.path.join(normal_path, f))]
covid_valid = all(f.startswith('COVID') for f in covid_files)
normal_valid = all(f.startswith('Normal') for f in normal_files)
print(f"Tous les fichiers dans COVID commencent par 'COVID' : {covid_valid}")
print(f"Tous les fichiers dans Normal commencent par 'Normal' : {normal_valid}")
X = []
y = []
covid_images_list = []  # Pour stocker temporairement les images COVID
normal_images_list = []  # Pour stocker temporairement les images Normal

# Charger toutes les images COVID
os.chdir(covid_path)
for i in tqdm(os.listdir()):
    img = cv2.imread(i)
    if img is not None:
        img = cv2.resize(img, (224, 224))
        X.append(img)
        label = i[0:1] if i.startswith('COVID') else 'C'  # Secours si nom incorrect
        y.append(label)
        print(label)  # Imite print(i[0:1]) de votre professeur
        covid_images_list.append(img)  # Stocker pour affichage

# Charger un échantillon aléatoire de 3 616 images Normal
os.chdir(normal_path)
normal_images_sample = random.sample(os.listdir(), 3616)
for i in tqdm(normal_images_sample):
    img = cv2.imread(i)
    if img is not None:
        img = cv2.resize(img, (224, 224))
        X.append(img)
        label = i[0:1] if i.startswith('Normal') else 'N'  # Secours si nom incorrect
        y.append(label)
        print(label)  # Imite print(i[0:1])
        normal_images_list.append(img)  # Stocker pour affichage

# Vérifier l'équilibre
covid_count = sum(1 for label in y if label == 'C')
normal_count = sum(1 for label in y if label == 'N')
print(f"Nombre d'images COVID après équilibrage : {covid_count}")
print(f"Nombre d'images Normal après équilibrage : {normal_count}")

# Afficher la liste complète des étiquettes
print(y)  # Comme dans le code de votre professeur

# Afficher 2 images COVID et 2 images Normal
plt.figure(figsize=(10, 5))
for i in range(2):
    plt.subplot(2, 2, i + 1)
    plt.imshow(cv2.cvtColor(covid_images_list[i], cv2.COLOR_BGR2RGB))  # Convertir BGR en RGB
    plt.title('COVID')
    plt.axis('off')
for i in range(2):
    plt.subplot(2, 2, i + 3)
    plt.imshow(cv2.cvtColor(normal_images_list[i], cv2.COLOR_BGR2RGB))  # Convertir BGR en RGB
    plt.title('Normal')
    plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Visualiser l'équilibre
import matplotlib.pyplot as plt
%matplotlib inline
plt.figure(figsize=(6, 4))
plt.bar(['COVID', 'Normal'], [covid_count, normal_count], color=['#1f77b4', '#ff7f0e'])
plt.title('Nombre d\'images par classe après équilibrage')
plt.xlabel('Classe')
plt.ylabel('Nombre d\'images')
for i, count in enumerate([covid_count, normal_count]):
    plt.text(i, count + 50, str(count), ha='center', va='bottom')
plt.show()
```
